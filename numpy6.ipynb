```python
import numpy as np

A = np.array([
    [4, 5],
    [6, 7]
])
det = np.linalg.det(A)
print(det)
```

```python
import numpy as np

A = np.array([
    [2, 0],
    [0, 5]
])

eigenvalues , eigenvectors = np.linalg.eig(A)
print(eigenvalues)
```
