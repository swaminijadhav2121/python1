```python
for i in range(1,51):
    if i % 2 != 0:
        continue    
    if i % 5 == 0:
        continue
    print(i)
```

```python
for i in range(1,101):
    if i % 7 == 0 and i % 9 == 0:
        break
    print(i)
```

```python
n = 10
while n >= 1:
    if n == 6:
        n -= 1
        continue
    print(n)
    n -= 1
```
