```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])


print(arr.reshape(2, 3))

# 2. Convert into 1D
arr = arr.flatten()
print("Flattened Array:", arr)

# 3. Create copy
arr = arr.copy()

# 4. Insert 25 at index 2
arr = np.insert(arr, 2, 25)
print("After Insert:", arr)

# 5. Append 70
arr = np.append(arr, 70)
print("After Append:", arr)

# 6. Delete element at index 0
arr = np.delete(arr, 0)
print("After Delete:", arr)

# 7. Sort final array
arr = np.sort(arr)
print("Final Sorted Array:", arr)

# 8. Final array properties
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data Type:", arr.dtype)
```

```python
import numpy as np

numbers = []

for i in range(5):
    num = int(input("Enter number: "))
    numbers.append(num)

arr = np.array(numbers)

print("Original Array:", arr)

sorted_arr = np.sort(arr)

print(sorted_arr)
print(np.max(arr))
print(np.min(arr))
print(np.sum(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.unique(arr))
print(np.count_nonzero(arr))

additional = int(input("Enter additional number: "))

position = np.searchsorted(sorted_arr, additional)

print("Insertion Position:", position)
```
