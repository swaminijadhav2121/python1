```python
#nested if
# Bank Loan

age = int(input("Enter Age: "))
salary = int(input("Enter Salary: "))

if age >= 21:
    if salary >= 25000:
        print("Loan Approved")
    else:
        print("Loan Rejected")
else:
    print("Loan Rejected")
```
