```python
import numpy as np

A = np.array([
    [1,1],
    [2,3]
])
B = np.array([7,17])
print(np.linalg.solve(A,B))
```

```python
import numpy as np

A = np.array([
    [4,0],
    [0,5]
])
egenvalues,egenvectors = np.linalg.eig(A)
print(egenvalues)
print(egenvectors)
```

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
])
print(np.linalg.matrix_rank(A))
```

```python
import numpy as np

A = np.array([
    [2, 1],
    [4, 2]
])
det = np.linalg.det(A)
print(det)
if det == 0:
    print("No inverse")
else:
    print(np.linalg.inv(A))
```
