```python
#Print numbers from 10 down to 1 in reverse order.
class num:
    def num(self):
        for i in range(10,0,-1):
            print(i)


obj=num()
obj.num()
```

```python
#Calculate and print the sum of the first n natural numbers
num=int(input("enter your number:"))
class NaturalNum:
    def sum(self,n):
        total = 0
        
        for i in range(1,n + 1):
            total = total + i
        print("sum=",total)


obj=NaturalNum()
obj.sum(num)
```

```python
#Calculate and print the factorial of a given number.
num=int(input("enter your number:"))
class FactorialNum:
    def factorial(self,n):
        fact = 1
        
        for i in range(1,n + 1):
            fact = fact * i
            print("factorial=",fact)


obj=FactorialNum()
obj.factorial(num)
```

```python
#Find and print the product of all digits of a given number.
num=int(input("enter number:"))
class DigitProduct:
    def product(self,n):
        product = 1
        while n > 0:
            digit = n % 10
            product = product * digit
            n = n // 10

        print("Product =", product)

obj=DigitProduct()
obj.product(num)

        
```

```python
#Count and print the total number of digits in a given number.
num=int(input("enter number:"))
class DigitCount:
    def count(self,n):
        count = 0
        while n > 0:
            n = n // 10
            count = count + 1

        print("digit =", count)

obj=DigitCount()
obj.count(num)

        
```

```python
#Reverse the given number and print the reversed value.
num=int(input("enter number:"))
class ReverseNum:
    def reverse(self,n):
        reverse = 0
        while n > 0:
            digit = n % 10
            reverse = reverse * 10 + digit
            n = n // 10

        print("reverse number =", reverse)

obj=ReverseNum()
obj.reverse(num)

        
```

```python
#Check whether the given number is a palindrome
num=int(input("enter number:"))
class PalindromeNum:
    def palindrome(self,n):
        original = n
        reverse = 0
        while n > 0:
            digit = n % 10
            reverse = reverse * 10 + digit
            n = n // 10
        if original == reverse:
            print("palindrom")
        else:
            print("not palindrom")

obj=PalindromeNum()
obj.palindrome(num)
```

```python
#Find and print the sum of digits of the given number.
num = int(input("Enter number: "))

class DigitSum:
    def __init__(self, n):
        self.n = n
        
    def sum(self, n):
        total = 0

        while n > 0:
            digit = n % 10
            total = total + digit
            n = n // 10

        print("Sum =", total)


obj = DigitSum(num)
obj.sum(num)
```

```python
num = int(input("Enter the number: "))

class ArmstrongNum:
    def __init__(self, n):
        self.n = n

    def armstrong(self):
        original = self.n
        total = 0
        digits = len(str(self.n))

        while self.n > 0:
            digit = self.n % 10
            total = total + digit ** digits
            self.n = self.n // 10

        if original == total:
            print("Armstrong Number")
        else:
            print("Not Armstrong Number")


obj = ArmstrongNum(num)
obj.armstrong()
```
