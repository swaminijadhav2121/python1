```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(person):
    def __init__(self, name, age, roll_no):
        super().__init__(name,age)
        
        self.roll_no = roll_no
        
student = Student("swamini",21,102)

print("Name",student.name)
print("Age",student.age)
print("Roll No",student.roll_no)
```

```python
class Father:
    def father_property(self):
        print("father property")
class Mother:
    def mother_property(self):
        print("mother property")

class Child(Father , Mother):
    def child_activity(self):
        print("child activity")

child = Child()
child.father_property()
child.mother_property()
child.child_activity()
```

```python
class Person:
    def show_name(self):
        print("show name")

class Employee(Person):
    def show_salary(self):
        print("show salary")

class Manager(Employee):
    def show_department(self):
        print("show department")
        
manager = Manager()
manager.show_name()
manager.show_salary()
manager.show_department()
```

```python
class Employee:
    def login(self):
        print("employee login")

class Developer(Employee):
    def write_code(self):
        print("developer write a code")

class Tester(Employee):
    def test_software(Self):
        print("tester testing a software")

developer = Developer()
tester = Tester()

developer.login()
developer.write_code()

tester.login()
tester.test_software()

    
```

```python
class Animal:
    def eat(self):
        print("animals eat")

class Dog(Animal):
    def bark(self):
        print("dog bark")
        
class Cat(Animal):
    def meow(self):
        print("cat meow")

class Puppy(Dog,Cat):
    def play(self):
        print("puppy play")
        
puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.meow()
puppy.play()
```
