```python
import numpy as np

marks = np.array([35, 78, 42, 29, 65, 18, 90, 55])

passing = marks[marks >= 40]
failed = marks[marks < 40]

print("Passing marks:", passing)
print("Failed marks:", failed)
print("Number of passed students:", len(passing))
print("Number of failed students:", len(failed))
```
